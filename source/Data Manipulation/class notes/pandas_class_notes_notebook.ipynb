## Create DataFrames & read JSON/CSV

````python
data = {
    "id":   [1,2,3,4,5],
    "name": ["Ada","Alan","Grace","Linus","Edsger"],
    "score":[9.7, 8.2, 9.6, 7.5, 8.9],
    "program": ["CS","CS","DS","SE","CS"]
}
df = pd.DataFrame(data)
df
````

### read_json / read_csv

````python
students = [
    {"id": 101, "name": "Ada Lovelace",  "program":"CS", "score": 9.5},
    {"id": 102, "name": "Alan Turing",   "program":"CS", "score": 8.8},
    {"id": 103, "name": "Grace Hopper",  "program":"DS", "score": 9.6},
]
import json
from io import StringIO
js = json.dumps(students, ensure_ascii=False, indent=2)
dfj = pd.read_json(StringIO(js))
from io import StringIO as _SIO
csv_text = "id,name,comment,score\n1,Alan Turing,\"loves, commas\",10\n2,Grace Hopper,\"quotes \"\"are\"\" fine\",9.5\n"
dfc = pd.read_csv(_SIO(csv_text))
display(dfj, dfc)
````

## Inspect dtypes & options

````python
print(dfj.dtypes)
dfc.info()
````

## Selecting rows/columns (very brief)
- Column(s): `df['name']`, `df[['id','name']]`
- Row slice: `df[1:4]` (by position)
- loc/iloc: `df.loc[row_labels, col_labels]`, `df.iloc[row_idx, col_idx]`

````python
print(df[["id","name"]])
print(df[1:4])
print(df.loc[df["program"]=="CS", ["name","score"]])
print(df.iloc[:3, :2])
````

## Querying: masks, .query, string ops, isin, between

````python
high = df[df["score"] >= 9]
cs = df.query("program == 'CS' and score >= 8.5")
names_with_a = df[df["name"].str.contains("a", case=False)]
subset = df[df["program"].isin(["CS","DS"]) & df["score"].between(8.5, 9.8)]
display(high, cs, names_with_a, subset)
````

### B-Quick checks (asserts)

````python
assert set(cs["program"]) == {"CS"}
assert high["score"].min() >= 9
assert subset["score"].between(8.5, 9.8).all()
````

# Takeaways
- **JSON**: great for nested data; ensure valid JSON (no comments/trailing commas); control with `indent`, `ensure_ascii`.
- **CSV**: plain tabular text; be explicit with delimiter/quoting; watch commas in text; use `newline=''` on write.
- **pandas**: read JSON/CSV, inspect dtypes, select, and **query** with masks or `.query()`; string helpers like `.str.contains` are key.

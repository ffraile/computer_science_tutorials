## Iterables II (Tuples, dictionaries, nested iterables, comprehensions)
1. Tuple cheatsheet:
    - parentheses ```()``` to define a tuple
    - immutable (cannot be changed)
    - useful for fixed collections of items
2. Dictionary cheatsheet:
  - curly brackets `{}` to define a dictionary
  - ```key:value``` pairs
  - ```key``` must be unique
  - ```values()``` method returns a list of all the values in the dictionary
  - ```keys()``` method returns a list of all the keys in the dictionary
  - Better than lists when elements have different meaning
  - Accessing values: `dictionary[key]` returns the value associated with the key
  - Faster lookups than lists
  - Iteration over keys: ```for key in dictionary:```
  - Iteration over key-value pairs: ```for key, value in dictionary.items():```
  - Iteration over values: ```for value in dictionary.values()```
3. Nested iterables cheatsheet:
  - iterables inside iterables
  - iterables can be of different types
  - useful for representing complex data structures
  - common patterns: list of lists, list of dictionaries, dictionary of lists
  - Indexing from outer to inner: `nested_iterable[i][j]` First index accesses the outer iterable, second index accesses the inner iterable
  - Iteration with nested loops: Iterate over outer iterable, then inner iterable
4. Comprehensions cheatsheet:
  - concise way to create iterables
  - list comprehensions: `[expression for item in iterable if condition]`
  - dictionary comprehensions: `{key_expression: value_expression for item in iterable if condition}`

## File Handling and file formats (CSV, JSON)
1. File handling cheatsheet:
    - open a file: `open(filename, mode)`
    - modes: 'r' (read), 'w' (write), 'a' (append)
    - read from a file: `file.read()`, `file.readline()`, `file.readlines()`
    - write to a file: `file.write(data)`, `file.writelines(list_of_data)`
    - close a file: `file.close()`
    - use `with` statement to automatically close files
    - example:

```python
with open('example.txt', 'r') as file:
     content = file.read()
print(content)
```

2. CSV file format cheatsheet:
    - Comma-Separated Values
    - Each line represents a row
    - Values in a row are separated by commas
    - Use `csv` module to read/write CSV files
    - Example to read CSV:
```python
import csv
with open('data.csv', 'r') as file:
     reader = csv.reader(file)
     for row in reader:
        print(row)
```
- Example to write CSV:

```python
import csv
with open('data.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Alice', 30, 'New York'])
```
3. JSON file format cheatsheet:
    - JavaScript Object Notation
    - Text format for storing structured data
    - Same structure as Python dictionaries/lists
    - Use `json` module to read/write JSON files
    - Example to read JSON:
```python
import json
with open('data.json', 'r') as file:
    data = json.load(file)
    print(data)
```
- Example to write JSON:

```python
import json
data = {'name': 'Alice', 'age': 30, 'city': 'New York'}
with open('data.json', 'w') as file:
    json.dump(data, file)
```

## Serial Communication
1. Serial communication basics:
    - Data is transmitted one bit at a time over a single communication line (USB, Bluetooth, etc.)
    -  In Python, use the ```pyserial``` library to handle serial communication:
        - install: `pip install pyserial`
        - import: `import serial`
    - Open a serial port:
        - `ser = serial.Serial(port, baudrate, timeout)`
        - `port`: COM port (e.g., 'COM3' on Windows, '/dev/ttyUSB0' on Linux)
        - `baudrate`: communication speed (e.g., 9600, 115200)
        - `timeout`: read timeout in seconds
    - Write data to the serial port:
        - `ser.write(data.encode())` (data must be bytes)
    - Read data from the serial port:
        - `data = ser.readline().decode().strip()`
    - Close the serial port:
        - `ser.close()`
    - In Arduino, use `Serial.begin(baudrate)` in `setup()` and `Serial.print()` or `Serial.println()` to send data.
2. Example code to send and receive data:
```python
import serial
import time
ser = serial.Serial('COM3', 9600, timeout=1)
time.sleep(2)  # wait for the serial connection to initialize
ser.write('Hello, Arduino!'.encode())
ser.flush()  # wait for the data to be sent
data = ser.readline().decode().strip()
print(f'Received: {data}')
ser.close()
```
```c
void setup() {
  Serial.begin(9600);
}
void loop() {
  if (Serial.available() > 0) {
    String data = Serial.readLine();
    Serial.println("Echo: " + data);
  }
}
```
3. Common issues:
    - Incorrect COM port
    - Mismatched baud rates
    - Serial port already in use
    - Data encoding/decoding errors

## Pandas Basics
1. Pandas Basics:
    - Installing Pandas (not needed in Colab):
        - `pip install pandas`
    - Importing Pandas:
        - `import pandas as pd`
    - Creating Pandas DataFrames:
        - From dictionaries: `df = pd.DataFrame({'col1': [1, 2], 'col2': [3, 4]})`
        - From lists of lists: `df = pd.DataFrame([[1, 2], [3, 4]], columns=['col1', 'col2'])`
        - From CSV files: `df = pd.read_csv('file.csv')`
        - From JSON files: `df = pd.read_json('file.json')`
    - Viewing DataFrames:
        - First rows: `df.head(n)` (first n rows)
        - Header info: `df.info()`
        - Summary statistics: `df.describe()`
2. Accessing and modifying data:
    - Accessing columns: `df['col1']` or `df.col1`
    - Accessing rows by index: `df.iloc[0]` (first row)
    - Accessing rows by label: `df.loc[0]` (row with index label 0)
    - Adding new columns: `df['new_col'] = df['col1'] + df['col2']`
3. Querying data:
    - Using boolean masks: `df[df['col1'] > 2]`
    - Using `.query()`: `df.query('col1 > 2 and col2 < 5')`
    - String operations: `df[df['col1'].str.contains('pattern')]`
    - Using `isin()`: `df[df['col1'].isin([1, 2, 3])]`
    - Using `between()`: `df[df['col1'].between(2, 5)]`

## Numpy Basics
1. Numpy Basics:
    - Installing Numpy (not needed in Colab):
        - `pip install numpy`
    - Importing Numpy:
        - `import numpy as np`
    - Creating Numpy arrays:
        - From lists: `arr = np.array([1, 2, 3])`
        - From zeros: `arr = np.zeros((3, 4))` (3 rows, 4 columns)
        - From ones: `arr = np.ones((2, 5))`
        - Identity matrix: `arr = np.eye(3)` (3x3 identity matrix)
        - From random values: `arr = np.random.rand(3, 3)`
        - From random integers: `arr = np.random.randint(low, high, size=(3, 3))`
        - Linearly spaced values: `arr = np.linspace(start, stop, num)`
        - Range of values: `arr = np.arange(start, stop, step)`
    - Properties of Numpy arrays:
        - rank (number of dimensions): `arr.ndim`
        - Shape (number of elements in each dimension): `arr.shape`
        - Size (Total number of elements): `arr.size`
2. Accessing and modifying elements:
    - Indexing (one index per dimension, separated by commas): `arr[0, 1]` (first row, second column)
    - Slicing (one slice per dimension, separated by commas): `arr[0:2, 1:3]` (rows 0-1, columns 1-2)
    - Masking (using boolean arrays to select elements): `arr[arr > 5]` (elements greater than 5 in a 1D array)
    - Modifying elements: `arr[0, 0] = 10`
    - Shallow copy vs deep copy:
        - Shallow copy: `arr2 = arr1` (both refer to the same data)
        - Deep copy: `arr2 = arr1.copy()` (independent copy)
3. Broadcasting:
    - Automatic expansion of arrays with different shapes during arithmetic operations
    - Example: adding a 1D array to each row of a 2D array
        - ```python
            arr2d = np.array([[1, 2, 3], [4, 5, 6]])
            arr1d = np.array([10, 20, 30])
            result = arr2d + arr1d  # arr1d is broadcasted to match arr2d shape
          ```
4. Common Numpy functions:
    - Mathematical aggregations (optional axis parameter): Sum elements in array `np.sum(arr)`, Mean of elements in array `np.mean(arr)`, Standard deviation of elements `np.std(arr)`, Minimum of elements `np.min(arr)`, Max o elements `np.max(arr)`
        - Example with axis:
            - `np.sum(arr, axis=0)` (sum of each column)
            - `np.mean(arr, axis=1)` (mean of each row)
    - Element-wise operations:
        - Addition: `arr1 + arr2`
        - Subtraction: `arr1 - arr2`
        - Multiplication: `arr1 * arr2`
        - Division: `arr1 / arr2`
        - Exponentiation: `arr1 ** arr2`
        - Functions: Sine `np.sin(arr)`, Cosine `np.cos(arr)`, Exponential `np.exp(arr)`, Logarithm `np.log(arr)`, Square root `np.sqrt(arr)`
    - Reshaping arrays: `arr.reshape(new_shape)`
    - Transposing arrays: `arr.T`

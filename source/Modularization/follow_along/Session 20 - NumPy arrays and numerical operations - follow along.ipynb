## Scenario — Planning a small production run

In this scenario we are going to help the operations manager of a small workshop that manufactures two products, *Product A* and *Product B*. Every week, the manager prepares a **production plan**: the number of units of each product that the workshop intends to make. At the end of the week, the plan is compared with the **actual production**, the number of units that were really made.

This week, the plan is 100 units of Product A and 60 units of Product B, while the workshop actually produced 96 and 64 units. Making each unit consumes **resources**: two raw materials (Material X and Material Y, measured in kilograms) and machine time (measured in hours). The amount of each resource needed to make one unit of a product is called its **requirement**.

The manager wants to:

- compare planned and actual quantities;
- calculate deviations without processing every product manually;
- represent material requirements in a table;
- calculate the total resources required by a production plan;
- visualise the plan and actual output.

With two products we could do every calculation by hand, but a real factory may have hundreds of products and resources. We need a tool that applies the same calculation to many numbers at once, and that tool is NumPy.

This context is different from the sensor grid used in the opening activity, but it uses the same array ideas.

## 1. Import NumPy

Recall that a **library** is a collection of code written by other people that we can import into our programs. **NumPy** (*Numerical Python*) is an external library for numerical computing: it provides a fast data structure for numbers and many mathematical operations over it. Most scientific and data libraries in Python, including Pandas and Matplotlib, are built on top of NumPy.

The conventional **alias** is `np`: a short alternative name given with `as`, so that we can write `np.array()` instead of `numpy.array()`.

```python
import numpy as np
```

## 2. Compare a list with an array

A NumPy **array** is a collection of elements, like a list, with two important differences:

- all its elements have the same type, for instance all integers or all decimal numbers;
- arithmetic operators such as `+`, `-`, `*` and `/` act on the numbers stored in the array, not on the collection as a whole.

We create an array with `np.array()`, passing a list of values. Recall that, for lists, `+` joins two lists and `*` with an integer repeats the list.

Predict both results before running the cell.

````python
planned_list = [100, 60]
planned_array = np.array([100, 60])

print("List multiplied by 2:", planned_list * 2)
print("Array multiplied by 2:", planned_array * 2)
````

## 3. Create arrays and inspect their properties

An array can have one or more **dimensions**, or axes. We use three names from linear algebra:

- a **scalar** is a single number, with no dimensions;
- a **vector** is a one-dimensional array: a row of numbers, such as the planned quantities of the two products;
- a **matrix** is a two-dimensional array: a table of numbers organised in rows and columns.

Every array stores several properties as attributes, which we read with a dot:

| Property | Meaning | Example |
|---|---|---|
| `ndim` | Number of dimensions | `1` for a vector |
| `shape` | Number of positions along each dimension | `(2,)` |
| `size` | Total number of elements | `2` |
| `dtype` | Common storage type | integer or floating point |

Notice that `shape` is a tuple with one element per dimension. `(2,)` is a tuple with a single element: the trailing comma distinguishes it from the number `2` written between parentheses. The `dtype` (*data type*) is usually `int64` for integers and `float64` for decimal numbers; the number `64` is the number of bits used to store each element.

Predict these properties before running the cell.

````python
planned = np.array([100, 60])
actual = np.array([96, 64])

print("Planned:", planned)
print("Dimensions:", planned.ndim)
print("Shape:", planned.shape)
print("Size:", planned.size)
print("Data type:", planned.dtype)
````

## 4. Apply vectorised operations

A **vectorised operation** applies an operation to every array element without an explicit Python loop. When two arrays with the same shape are combined with an arithmetic operator, the operation is **element-wise**: NumPy combines the element at position `0` of the first array with the element at position `0` of the second one, then position `1` with position `1`, and so on. The result is a new array with the same shape.

When one of the operands is a scalar, as in `actual / planned * 100`, the scalar is applied to every element.

The deviation for each product is:

```text
actual quantity − planned quantity
```

Without NumPy, we would need a loop and an accumulator:

```python
deviation = []
for index in range(len(actual_list)):
    deviation.append(actual_list[index] - planned_list[index])
```

With NumPy, one expression says the same thing. Predict the result and its shape.

````python
deviation = actual - planned
percentage_achieved = actual / planned * 100

print("Deviation:", deviation)
print("Percentage achieved:", percentage_achieved)
print("Deviation shape:", deviation.shape)
print("Percentage data type:", percentage_achieved.dtype)
````

### Interpret, do not only calculate

- A negative deviation means production was below plan.
- A positive deviation means production exceeded plan.
- Division can change the resulting `dtype` even when both inputs contain integers.

Explain why `percentage_achieved` is a floating-point array and why its shape remains `(2,)`.


## 5. Reuse array logic inside a function

Functions and vectorisation work together. The function below receives two compatible arrays and returns a new array.

Before running it, identify its inputs, output and numerical rule.


````python
def calculate_percentage_achieved(actual_values, planned_values):
    """Return the percentage achieved for every planned quantity."""
    return actual_values / planned_values * 100

achievement = calculate_percentage_achieved(actual, planned)
print(achievement)
````

## 6. Create a two-dimensional array

A list of lists can become a two-dimensional array. Each inner list becomes one row of the matrix, so all inner lists must have the same length. Here, rows represent resources and columns represent products.

| Resource | Product A | Product B |
|---|---:|---:|
| Material X (kg/unit) | 2.0 | 1.5 |
| Material Y (kg/unit) | 0.5 | 0.8 |
| Machine time (h/unit) | 1.0 | 1.2 |

For instance, the value `1.5` in the first row and second column means that one unit of Product B needs 1.5 kg of Material X.

Predict `ndim`, `shape`, `size` and `dtype`.

````python
requirements = np.array([
    [2.0, 1.5],
    [0.5, 0.8],
    [1.0, 1.2],
])

print(requirements)
print("Dimensions:", requirements.ndim)
print("Shape:", requirements.shape)
print("Size:", requirements.size)
print("Data type:", requirements.dtype)
````

## 7. Select elements, rows, columns and regions

In a two-dimensional array, we write both indexes inside one pair of square brackets, separated by a comma. The first index selects a row and the second selects a column.

```python
array[row_index, column_index]
```

A colon on its own selects every position in a dimension, so `requirements[0, :]` reads as *row 0, all columns*. A slice keeps the familiar rule from session 18: the start is included and the stop is excluded. A rectangular part of a matrix selected with two slices is called a **region** (or sub-matrix).

Predict the value and shape of every result below.

````python
one_value = requirements[1, 0]
first_resource = requirements[0, :]
product_b_requirements = requirements[:, 1]
selected_region = requirements[1:3, 0:2]

print("One value:", one_value)
print("First resource row:", first_resource, first_resource.shape)
print("Product B column:", product_b_requirements, product_b_requirements.shape)
print("Selected region:\n", selected_region, selected_region.shape)
````

### Explain the dimensional change

Why does `requirements[1, 0]` return one scalar value while `requirements[1:2, 0:1]` returns a two-dimensional array? The selected numbers can be similar, but the selections preserve different structures.


## 8. Element-wise operations

The operator `*` multiplies corresponding elements: the value in row `i` and column `j` of the first array is multiplied by the value in row `i` and column `j` of the second one. This requires compatible element arrangements, so first we compare two arrays with the same shape.

In `cost_per_unit`, each row contains the cost of one unit of a resource, in euros: 3 € per kg of Material X, 5 € per kg of Material Y and 20 € per machine hour. The value is repeated in both columns because the cost of a resource does not depend on the product.

````python
cost_per_unit = np.array([
    [3.0, 3.0],
    [5.0, 5.0],
    [20.0, 20.0],
])

cost_requirements = requirements * cost_per_unit
print(cost_requirements)
print("Result shape:", cost_requirements.shape)
````

### What question did `*` answer?

The result still contains one value for every original matrix position. It answers: “What is the cost contribution of this resource for one unit of this product?” It does **not** yet calculate totals for the complete production plan.


## 9. Matrix multiplication combines rows with a vector

The operator `@` performs **matrix multiplication**, a different operation from `*`. To understand it, we first need the **dot product** (or scalar product) of two vectors with the same length: multiply corresponding elements and add the products. For instance, the dot product of `[2.0, 1.5]` and `[100, 60]` is `2.0 × 100 + 1.5 × 60`.

In the product of a matrix and a vector, each element of the result is the dot product of one row of the matrix with the vector. For this to be possible, the length of each row (the number of columns of the matrix) must be equal to the length of the vector. These two numbers are called the **inner dimensions**, and the shape of the result is given by the **outer dimensions**:

```text
(3, 2) @ (2,) → (3,)
```

Here, each resource row is combined with the two planned product quantities, so the result has one value per resource.

Trace the first result manually:

```text
2.0 × 100 + 1.5 × 60 = ?
```

````python
total_requirements = requirements @ planned

print("Requirements shape:", requirements.shape)
print("Plan shape:", planned.shape)
print("Total requirements:", total_requirements)
print("Result shape:", total_requirements.shape)
````

### Interpret the result

The three result positions represent:

1. total kilograms of Material X;
2. total kilograms of Material Y;
3. total machine hours.

Matrix multiplication is appropriate because the program combines every resource row with the complete production vector.


## 10. Transpose changes the orientation

The **transpose** of a matrix exchanges its rows and columns: the first row becomes the first column, the second row becomes the second column, and so on. In NumPy, the attribute `.T` returns the transposed array. A matrix with shape `(3, 2)` becomes a matrix with shape `(2, 3)`, and the element in row `i` and column `j` moves to row `j` and column `i`.

The transpose does not change the total number of elements. Predict the new shape before running the cell.

````python
requirements_transposed = requirements.T

print(requirements_transposed)
print("Original shape:", requirements.shape)
print("Transposed shape:", requirements_transposed.shape)
print("Original size:", requirements.size)
print("Transposed size:", requirements_transposed.size)
````

### Dimensional checkpoint

Decide whether each expression is dimensionally compatible before trying it:

1. `requirements @ planned` with `(3, 2) @ (2,)`
2. `requirements.T @ planned` with `(2, 3) @ (2,)`
3. `requirements.T @ total_requirements` with `(2, 3) @ (3,)`

Only run an expression after explaining whether the inner dimensions match.


````python
# Run only the compatible expressions after discussing them.
print("1:", requirements @ planned)
# print("2:", requirements.T @ planned)
print("3:", requirements.T @ total_requirements)
````

## 11. Plot arrays with Matplotlib

Matplotlib accepts NumPy arrays directly. The chart below compares planned and actual production with a **bar chart**, which represents each value as a bar whose height is proportional to the value.

Recall `plt.plot()` from Unit 02. `plt.bar(x, height, width=..., label=...)` works in a similar way:

- `x` contains the horizontal position of each bar, and `height` contains its value;
- we draw two series of bars, so we shift the planned bars `0.15` to the left (`positions - 0.15`) and the actual bars `0.15` to the right. Notice that this is again a vectorised operation;
- `plt.xticks(positions, product_names)` replaces the numeric positions with the product names on the horizontal axis;
- `plt.legend()` displays the `label` of each series.

The labels remain ordinary strings; the numerical series are arrays.

````python
import matplotlib.pyplot as plt

product_names = ["Product A", "Product B"]
positions = np.array([0, 1])

plt.bar(positions - 0.15, planned, width=0.3, label="Planned")
plt.bar(positions + 0.15, actual, width=0.3, label="Actual")
plt.xticks(positions, product_names)
plt.ylabel("Units")
plt.title("Planned and actual production")
plt.legend()
plt.show()
````

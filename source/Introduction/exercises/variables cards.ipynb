### Card B-3 — Human Calculator
```python
def ping():
    print("ping")
    return True
print(False and ping())
print(True or ping())
```
What is printed?

````python

# Card B-3 — check (short-circuit)
def ping():
    print("ping")
    return True

print(False and ping())  # expected: False   (ping not called)
print(True or ping())    # expected: True    (ping not called)
````

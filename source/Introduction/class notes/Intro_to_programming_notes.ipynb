## What is a **program**?
A precise, step-by-step set of instructions that a computer can execute to transform inputs into outputs.

## What is a **programming language**?
A **language** (means of communication) with **formal notation** including **syntax** (how code is written) and **semantics** (what code means) that lets humans describe **programs** for machines.

---

## Syntactic Units
- **Expression:** produces a value. *(e.g., `2+3`, `len(name)`, `[1,2][0]`)*
- **Statement:** performs an action; may use expressions. *(e.g., `x = 2+3`, `print(x)`, `return y`)*
- **Block:** a group of statements controlled by structure/indentation. *(e.g., the body of a function, `if`, or `for`)*

> Mental model:
- **expressions** → **words**
- **statements** → **sentences**
- **blocks** → **paragraphs**.

---

## Variables
- **Definition:** a *name* bound to a *value/object* in memory.
- **Assignment:** `name = expression` binds/updates a value. It does **not** equal a math equation; it’s an action.

`name`in left hand side cannot contain spaces
`expression`in right hand side must be a valid expression

```python
x = 2              # assign 2 to name x
x = x + 3          # update: now x is 5
```

---

## Functions in Python
- **Function (idea):** a reusable block that takes **parameters** and may **return** a value.
- **Function call (action):** evaluate arguments → execute the function → (optionally) get a return value.
- **Built-in functions:** provided by Python, e.g., `print`, `len`, `sum`.

```python
def add(a, b):       # function *definition* (a, b are parameters)
    return a + b     # returns a value (no printing here)

result = add(2, 3)   # function *call* (2, 3 are arguments)
print(result)        # Built-in function: display to screen
```

### Function definition
- **Syntax:** `def`keyword + function name + parameter list + optional default values.
- **Basic form:** def name(param1, param2=default)
- **Advanced form:** def name(*args, **kwargs, param1, param2=default)
- **Examples:**
```python
def fib(n): ...
def switch_led(mode="blink", color="red"): ...
```

---

### Function calls
- **Parameters**: names in the function definition (`a`, `b`).
- **Arguments**: values you pass at the call site (`add(2, 3)`).
- **Types**:
  - keyword: Uses parameter name
  - positional: not using name, argument position determines parameter.
  - variadic: number of parameters can vary (`*args` - positional, `**kwargs`- keyed).
```python
fib(n=3) # keyword parameter
fib(5)   # positional parameter
fib()    # Raises error: n does not have a default value
switch_led() # both parameters apply default value
switch_led(color="blue") # mode parameter applies default value
switch_led("static", "blue") #positional: mode takes value "static, color value "blue"
print("Hello", "World")      # varying number of arguments
```

###  Scope & Lifetime

- **Global variables**: Defined in "main program"
- **Local variables**: Defined inside functions. They are only accesible locally inside the function

```python
x = 3 # Global variable
def eq():
  y = 2
  z = x*y
  return z

print(x) # This works
print(y) # This fails. y is only accessible in the local context of eq
```

## What is a **program**?
A precise, step-by-step set of instructions that a computer can execute to transform inputs into outputs.

## What is a **programming language**?
A **language** (means of communication) with **formal notation** including **syntax** (how code is written) and **semantics** (what code means) that lets humans describe **programs** for machines.

---

## Syntactic Units
- **Expression:** produces a value. *(e.g., `2+3`, `len(name)`, `[1,2][0]`)*
- **Statement:** performs an action; may use expressions. *(e.g., `x = 2+3`, `print(x)`, `return y`)*
- **Block:** a group of statements controlled by structure/indentation. *(e.g., the body of a function, `if`, or `for`)*

> Mental model:
- **expressions** → **words**
- **statements** → **sentences**
- **blocks** → **paragraphs**.

---

## Variables
- **Definition:** a *name* bound to a *value/object* in memory.
- **Assignment:** `name = expression` binds/updates a value. It does **not** equal a math equation; it’s an action.

`name`in left hand side cannot contain spaces
`expression`in right hand side must be a valid expression

```python
x = 2              # assign 2 to name x
x = x + 3          # update: now x is 5
```

---

## Functions & Function Calls
- **Function (idea):** a reusable block that takes **parameters** and may **return** a value.
- **Function call (action):** evaluate arguments → execute the function → (optionally) get a return value.
- **Built-in functions:** provided by Python, e.g., `print`, `len`, `sum`.

```python
def add(a, b):       # function *definition* (a, b are parameters)
    return a + b     # returns a value (no printing here)

result = add(2, 3)   # function *call* (2, 3 are arguments)
print(result)        # side-effect: display to screen
```

> Keep core logic **pure** (use `return`) and print at the edges.

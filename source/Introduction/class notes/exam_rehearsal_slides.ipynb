## Unit 1 — Processor/Registers/Control (Rehearsal)

### U1.R1 — Decode

Processor uses the following operation format: **16-bit** `[OPC(4) | RD(4) | RO1(4) | RO2(4)]`.
Where:
- `OPC(4)` is the 4-bit operation code
- `RD(4)` is the 4-bit destination registry identifier
- `RO1(4)` is the 4-bit first operand
- `RO2(4)` is the second 4-bit second operand

Values in the 4-bit registries are **signed-extended** meaning that values are in the range `[-8,7]` with the most significant bit (MSB) representing the sign (`1` for negative numbers and `0` for positive numbers). For instance (`0b0100` is decoded as `7`, `0b1000` is decoded as `-8`, and `0b1111` is decoded as `-1`)

**Task 1:**
1) Parse bits of instruction: `0x7D3A` and decode `RD(4)`, `RO1`, and `RO2`.


````python

# U1.R1 — run to check
word = 0x7D3A
print(f"the word in binary format is {bin(word)}")  #bin function returns binary representation of number
# We can also parse the values using bitwise: right shift and bitwise and
# WARN: Bitwise operations not covered in the exam! Covered in the rehearsal for practice
OPC = (word >> 12) & 0xF
RD  = (word >> 8)  & 0xF
RO1  = (word >> 4)  & 0xF
RO2 =  word        & 0xF

def decode(x):
    return x if x < 8 else x - 16  # sign-extend 4-bit to Python int
print(f"RD: {decode(RD)}, RO1: {decode(RO1)}, RO2: {decode(RO2)}")
````

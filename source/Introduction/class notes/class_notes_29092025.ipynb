### 4.3 Logical Operators:
  - `and` (alias `&`) returns True if both operands are True (`a and b` `a & b`)
    - **short-circuit**: if first operand is False, second is not computed
  - `or` (alias `|`) returns True if either operand is True (`a or b` `a | b`)
    - **short-circuit**: if first operand is True, second is not computed
  - `not' flips the result (`not True` yields `False` and `not False` yields `True`).
  - `xor` (aliases `!=`, `^`) returns True if exactly one operand is True
    - `xor` is not a keyword, just the name of the operation, but `!=` and `^` can be used
    - No short-circuit: both operands are always evaluated

## 1) Basic Control Structures

### 1.1 `if / else` (branching or Selection)

**Definition:** Execute a block **only if** a condition is `True`.
**Syntax:**
```python
if condition:
    # block A (executes when condition is True)
else:                 # optional
    # block B (executes when condition is False)
```
**Important**
- **Indentation defines the block.**
- The condition must evaluate to a boolean (or truthy/falsey).

````python

# Example: simple thermostat decision
temp = 18
if temp < 19:
    action = "HEAT_ON"
else:
    action = "COMFY_OR_COOL"
print(action)
````

### 1.2 `elif` (multiple branches)

**Definition:** Chain mutually‑exclusive conditions in priority order.
**Syntax:**
```python
if cond1:
    ...
elif cond2:
    ...
else:
    ...
```
**Important**
- Conditions are checked **top‑down**; first `True` branch runs; the rest are skipped.
- Boundaries matter—use `<=`/`>=` deliberately.

> ☝ Be aware of indentation! Indented code will be executed if the condition is met, so you need to be careful and indent only the right number of lines

❓**Predict then run**: Think about what the following code prints:

````python

# Example: check if odd even or zero
n = int(input("Enter a number: "))
if n == 0:
    print("You entered 0")
elif n % 2 == 0:
    print("You entered an even number ")
else:
    print("You entered an odd number ")
````

### 1.3 Compound conditionals

Combine conditions with boolean operators. Use **parentheses** for clarity.

````python

x = 3
y = 5
if (x == 3) and (y == 5):
    print("Both match")
if (x == 3) or (y == 4):
    print("At least one condition is true")
````

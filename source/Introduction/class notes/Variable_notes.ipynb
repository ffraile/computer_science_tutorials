## 0) (Reminder) Variables & Assignment
- **Variable**: human readable *name* givent to *object in memory*.  
- **Assignment**: `name = expression` (an action, not a math equation).
- Inspect: `type(x)` → let us know the type of a variable.

```python
x = 2
x = x + 3     # now x is 5
type(x)       # <class 'int'>
```

---

## 1) Methods and Attributes
- **Methods**: Functions that are associated with the type of the object, and are called using the dot (.) operator on the variable.

- **Attributes:** Variables that are associated with the type of the object, and are accessed using the dot (.) operator on the variable.

## 2) Numbers (int, float, complex)
### 2.1) Numeric types
- Integer (`int`): Integer numbers, i.e. numbers with no decimal part
- Float (`float`): Numbers with fractional part
- Complex (`complex`): Numbers with real and imaginary part
### 2.2 Arithmetic operators
- Perform mathematical operations on numerical values: `+ - * / // % **`
  - ```+``` addition (eg. ```3+2``` → 5)
  - ```-``` substraction (eg. ```3-2``` → 1)
  - ```*``` multiplication (eg. ```3*2``` → 6)
  - ```/``` division (eg ```3/2```→ 1.5)
  - ```//``` Integer division: Integer part of the division (eg ```3//2```→ 1)
  - ```%``` Modulo (remainder after integer division (eg ```3%2``` → 1)
  - ```**``` Exponentiation (eg ```3**2``` → 9)
- Precedence matters (use parentheses!). **Precedence levels**
  - Parentheses ()
  - Exponentiation ```**```
  - Multiplication ```*```, Division ```/```, Floor Division ```//```, Modulus ```%```
  - Addition ```+```, Subtraction ```-```
  - Arithmetic built-ins: `abs`, `round`, `min`, `max`.

Expressions within parentheses are evaluated first. Operations at the same level are evaluated from left to right.

```python
7 / 2   # 3.5
7 // 2  # 3
7 % 2   # 1
(1 + 2) * 3 / 4 - 0.5
```

Some attributes and methods
```python
2.25.as_integer_ratio # (9, 4)
x = 3 + 4*j # x is a complex number
x.real # 3
x.imag # 4
```

## 3) Strings (text)
- Literals:
  - Single or double quotes for single line `'...'`, `"..."`,
  - Triple double-quote for multiline `"""..."""`.  
- Strings are **immutable** (operations and methods return new strings).
- Basic operations:
  - `+` concatenates strings (`"a"+"b"+"c"` → `"abc"`)
  - `*n` repeats a string (`"a"*3` → `"aaa"`
- Useful methods:
  - `.upper()` converts to upper letter (`'fyi'.upper()` → `'FYI'`)
  - `.lower()` converts to lower leters (`'FYI'.lower()` → `'fyi'`)
  - `.replace()` replaces a substring with another (`'LUCK'.replace('L','#')` → `'#UCK'`)
  - `.count()` counts number of characters a substring appears in another (`'LUCK'.count('L')` → `'1'`)
  - `.find()` finds first apperance of substring (`"LUCK".find("CK")` → `2`)
  - `.startswith()` tells whether string starts with a substring (`"LUCK".startswith("F") → `False`)
  - `.endswith()` tells whether string ends with a substring (`"LUCK".endswith("F") → `False`)

### Formatting:
- Formatting allows us to format variables into a human readable form
 - `str.format(...)` or **f-strings** (preferred).

```python
name = input("What is your name?")
g1 = "Hello, " + name + "!"
g2 = "Hello, {}!".format(name)
g3 = f"Hello, {name}!"

pi = 3.14159
g3 = f"π≈{pi:.2f}"    # (2 decimals in floating format π≈3.14)
g4 = f"{1/3:.1%}".    # (1 decimal in percentage format 33.3)
```

## 4) Booleans
- Can only take two values (True/False)
- Logical Operators:
  - `and` returns True if both operands are True (`a and b` `a & b`)
    - **short-circuit**: if first operand is False, second is not computed
  - `or` returns True if either operand is True (`a or b` `a | b`)
    - **short-circuit**: if first operand is True, second is not computed
  - `not' flips the result (`not True` yields `False` and `not False` yields `True`).
    - `!a` is equivalent to `not a`
  - `xor` returns True if exactly one operand is True
- Comparison operators:
 - `==` exactly equal (e.g. `2 == 3` → `False`)
 - `!=`unequal (e.g. `2 != 3` → `True`)
 - `<`lower (e.g. `2 < 3` → `True`)
 - `<=`lower or equal (e.g. `2 <= 3` → `True`)
 - `>`greater (e.g. `2 > 3` → `False`)  
 - `>=` greater or equal (e.g. (e.g. `2 >= 3` → `False`)   

> ⚠️ Floating-point: `0.1 + 0.2` ≠ `0.3` exactly → `round(0.1 + 0.2, 2)`.
- Logic: `and`, `or`, `not`  (short-circuit)
- **Truthiness** (falsey): `0`, `0.0`, `""`, `[]`, `{}`, `None`

## 7) Common Pitfalls (call them out)
- Using `is` where `==` is intended.  
- Forgetting parentheses → precedence surprises.  
- Expecting string methods to modify in place (they return new strings).  
- Treating `/` like floor division (use `//` if you need floor).

**Takeaway:** Names bind to objects; know your core types; prefer explicit conversions & parentheses; format with f-strings; test truthiness carefully.

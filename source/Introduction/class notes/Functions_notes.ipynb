## 1) Function Signature
- **Signature:** function name + parameter list + optional default values.
- **Basic form:** def name(param1, param2=default)
- **Advanced form:** def name(*args, **kwargs, param1, param2=default)
- **Examples:**
```python
def fib(n): ...
def switch_led(mode="blink", color="red"): ...
```

---

## 2) Function calls
- **Parameters**: names in the function definition (`a`, `b`).
- **Arguments**: values you pass at the call site (`add(2, 3)`).
- **Types**:
  - keyword: Uses parameter name
  - positional: not using name, argument position determines parameter.
  - variadic: number of parameters can vary (`*args` - positional, `**kwargs`- keyed).
```python
fib(n=3) # keyword parameter
fib(5)   # positional parameter
fib()    # Raises error: n does not have a default value
switch_led() # both parameters apply default value
switch_led(color="blue") # mode parameter applies default value
switch_led("static", "blue") #positional: mode takes value "static, color value "blue"
print("Hello", "World")      # varying number of arguments
```

## 4) Scope & Lifetime

- **Global variables**: Defined in "main program"
- **Local variables**: Defined inside functions. They are only accesible locally inside the function

```python
x = 3 # Global variable
def eq():
  y = 2
  z = x*y
  return z

print(x) # This works
print(y) # This fails. y is only accessible in the local context of eq
```

- **Nested functions**: A function that is *local* to another, only accessible in the context of the outer function

```python
def outer_eq():
  x = 3
  def inner_eq():  # inner function, we can only use it locally
    y = 2          #y is only accessible in the context of the inner function
    return x + y
  return inner_eq() + 2

z = outer_eq()
print(z)
print(inner_eq())   #error: inner function is not callable in this context
```  

- **LEGB** (name resolution order)
  - **L**ocal: names assigned inside the current function.
  - **E**nclosing: names in outer (non-global) functions (closures).
  - **G**lobal: module-level names.
  - **B**uilt-ins: names like `len`, `print`.

```python
x = 10               # global
def outer():
    x = 20           # enclosing
    def inner():
        x = 30       # local
        return x
    return inner()
print(outer())       # 30; LEGB picks local first
print(x)             # 10; global unchanged
```

- **Write access rules:** assigning to a name makes it **local** unless declared `global` or `nonlocal`.

Global declaration: Forces use of global value
```python
count = 0
def bump():
    global count
    count += 1
```
Nonlocal: Forces use of outer context value

```python
def outer():
    total = 0
    def add1():
        nonlocal total
        total += 1
    add1()
    return total
```

## 5) Pure vs Side-Effectful
- **Pure function:** output depends only on inputs; no external state changes.
  - ✅ Easier to test, reason about, and reuse.
- **Side-effectful:** prints, writes files, mutates globals/arguments, I/O.

```python
def pure_add(a, b):           # pure
    return a + b

log = 0
def impure_add(a, b):         
    s = a + b
    print(s)                  # side effect: I/O
    log += s                  # side effect: modifies external variable
    return s
```
**Guideline:** keep **core logic pure**; perform I/O at the "edges".

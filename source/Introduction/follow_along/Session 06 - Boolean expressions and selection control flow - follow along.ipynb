## Scenario: event check-in

A community innovation hub is hosting a workshop. Its check-in program must decide whether a person can enter, needs to register, or should join a waiting list.

The program will use registration status, staff status, account restrictions and current occupancy. This is a new context: the same programming concepts can support decisions in many domains.

## 1. Comparisons produce Boolean values

A **comparison** evaluates two operands and produces one of the Boolean values `True` or `False`.

| Operator | Meaning | Example |
|---|---|---|
| `==` | equal to | `role == "staff"` |
| `!=` | not equal to | `role != "guest"` |
| `<` | less than | `attendees < capacity` |
| `<=` | less than or equal to | `queue_size <= 10` |
| `>` | greater than | `waiting_time > 15` |
| `>=` | greater than or equal to | `age >= 18` |

Predict every result before running the next cell.

````python
attendees = 29
capacity = 30
minimum_age = 18
visitor_age = 18
role = "participant"

print(attendees < capacity)
print(attendees == capacity)
print(visitor_age >= minimum_age)
print(role != "guest")
````

## 2. Boolean variables and program state

A Boolean result can be stored in a variable. A clear name records what the result means in the current program state.

````python
has_space = attendees < capacity
is_adult = visitor_age >= minimum_age
is_registered = True
account_blocked = False

print("Space available:", has_space)
print("Adult visitor:", is_adult)
print(type(has_space))
````

## 3. `if`: execute an optional branch

An `if` statement executes its indented block only when its condition is `True`. If the condition is `False`, the block is skipped.

```python
if condition:
    instruction
```

Indentation shows which instructions belong to the branch.

````python
check_in_message = "Continue with the standard check-in"

if attendees >= capacity:
    check_in_message = "The room is full"

print(check_in_message)
````

Change `attendees` first to `30` and then to `31`. Before each execution, predict whether the assignment inside the `if` block will run.

## 4. `if-else`: select one of two branches

An `if-else` statement always selects exactly one branch. The `if` block runs when the condition is `True`; otherwise, the `else` block runs.

````python
attendees = 29
capacity = 30

if attendees < capacity:
    capacity_message = "A place is available"
else:
    capacity_message = "Join the waiting list"

print(capacity_message)
````

Which branch should run when `attendees` is exactly equal to `capacity`? Explain how the operator encodes that rule.

## 5. `and`: require both conditions

The operator `and` produces `True` only when both operands are `True`. The visitor must be registered **and** the room must have space.

````python
is_registered = True
has_space = attendees < capacity
can_use_standard_check_in = is_registered and has_space

print("Registered:", is_registered)
print("Space available:", has_space)
print("Standard check-in allowed:", can_use_standard_check_in)
````

## 6. `or`: accept alternative conditions

The operator `or` produces `True` when at least one operand is `True`. A person has recognised access when they are registered **or** they are a staff member.

````python
is_registered = False
is_staff = True
has_recognised_access = is_registered or is_staff

print("Recognised access:", has_recognised_access)
````

## 7. `not`: reverse a Boolean value

The operator `not` reverses one Boolean value. `not account_blocked` is `True` only when the account is not blocked.

````python
account_blocked = False
account_allowed = not account_blocked

print("Blocked:", account_blocked)
print("Allowed by account status:", account_allowed)
````

## 8. Build a compound condition

A compound condition combines smaller Boolean expressions. Parentheses make the intended groups easier to review.

Read the next condition from the inside out:

1. the person is registered or is staff;
2. the account is not blocked;
3. the room has space;
4. all three requirements must be true.

````python
attendees = 29
capacity = 30
is_registered = False
is_staff = True
account_blocked = False

can_enter = (
    (is_registered or is_staff)
    and not account_blocked
    and attendees < capacity
)

print("Access granted:", can_enter)
````

Change one input at a time. Which single changes make `can_enter` false? Why is changing one value at a time useful when testing a condition?

## 9. `if-elif-else`: prioritise several outcomes

Python evaluates the conditions from top to bottom. Only the first branch whose condition is `True` executes. The `else` branch runs only when every previous condition is `False`.

The order below encodes a policy: a blocked account takes priority over every other outcome.

````python
attendees = 30
capacity = 30
is_registered = True
is_staff = False
account_blocked = False

if account_blocked:
    status = "Access denied: blocked account"
elif not (is_registered or is_staff):
    status = "Registration required"
elif attendees >= capacity:
    status = "Join the waiting list"
else:
    status = "Access granted"

print(status)
````

## 13. Integrated check-in program

This version connects previous learning about `input()` and conversion with today's decisions. Enter `yes` or `no` exactly as shown. Predict the branch before providing the final input.

````python
visitor_name = input("Visitor name: ")
is_registered = input("Registered (yes/no): ") == "yes"
is_staff = input("Staff member (yes/no): ") == "yes"
account_blocked = input("Blocked account (yes/no): ") == "yes"
attendees = int(input("Current attendees: "))
capacity = int(input("Room capacity: "))

if account_blocked:
    status = "Access denied: blocked account"
elif not (is_registered or is_staff):
    status = "Registration required"
elif attendees >= capacity:
    status = "Join the waiting list"
else:
    status = "Access granted"

print(visitor_name + ": " + status)
````

## 14. Trueness for non-Boolean values
Non-Boolean values can be used in conditions. Python treats some values as `False` and others as `True`. The following are considered `False`:
- `None`: This keyword represents the absence of a value.
- zero of any numeric type, for example `0`, `0.0`
- empty strings, for example `""`
- empty iterables and collections (we will introduce them later), for example `[]`, `{}`, `()`

Contrary to that, the following are considered `True`:
- non-zero numbers, for example `1`, `-1`, `3.14`
- non-empty strings, for example `" "`, `"False"`, `"0"`
- non-empty iterables and collections, for example `[0]`, `{"key": "value"}`, `(None,)`

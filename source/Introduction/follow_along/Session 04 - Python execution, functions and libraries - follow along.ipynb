## 5. Built-in and library functions

Built-in functions are immediately available. Library functions belong to modules that must first be **imported**. The `math` module is included in Python's standard library, so it does not require a separate installation.

````python
import math

print(math.sqrt(81)) # sqrt returns the square root of a number
print(math.log2(8)) # log2 returns the base-2 logarithm of a number
````

| Expression | Origin | Import required? |
|---|---|---|
| `max(4, 9, 2)` | Built-in function | No |
| `math.sqrt(81)` | Standard-library module | Yes |
| `math.log2(8)` | Standard-library module | Yes |

Let us try to connect `math.log2(8)` with the binary representation of 8. What is the binary representation of 8? What is the base-2 logarithm of 8? How are they related?
**Tip**: Replace 8 with a different power of two, such as 4 or 16 and see if the relationship still holds.

````python
print(math.log2(8))
print(bin(8))
````

## 7. Three ways to import

Imports determine which names become available in the kernel. Compare the function call after each import form.

### 7.1. Import the entire module
The `math` module is imported, so the `sqrt()` function must be prefixed with `math.`.

````python
import math
print(math.sqrt(81))
````

### 7.2. Use an alias
An **alias** is a short name for a module. The `math` module is imported as `m`, so the `sqrt()` function must be prefixed with `m.`.

````python
import math as m  # m is an alias
print(m.log2(8))
````

### 7.3. Import specific functions
The `sqrt()` function is imported directly, so it can be called without a prefix.

````python
from math import ceil
print(ceil(4.2))
````

## Key Concepts
To sum up, this session introduced the following key concepts:
- A **function** is reusable code that performs a task.
- A **function call** asks Python to execute that function.
- A **literal** is a value written directly in source code.
- An **argument** is a value supplied to a function call.
- A **module** is a file containing Python code that can be imported.
- A **package** is a collection of modules that can be installed and imported.
- A **kernel** stores the current execution state: imports and other names created by executed cells.
- A **syntax error** occurs when Python cannot parse the code. A **runtime error** occurs when Python cannot execute the code.

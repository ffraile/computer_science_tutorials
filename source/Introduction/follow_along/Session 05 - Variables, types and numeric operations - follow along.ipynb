## 4. Naming variables

Variable names must start with a letter or underscore, followed by letters, digits or underscores.
They cannot contain spaces or hyphens, and they cannot be Python keywords.

A valid name is not necessarily a useful name. Prefer descriptive `snake_case` names.

## 7. Applied example: time conversion

A process has been running for 10,000 seconds. We need complete hours, complete minutes and remaining seconds. Why do we need both `//` and `%`?

````python
total_seconds = 10_000

hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print(hours, "hours")
print(minutes, "minutes")
print(seconds, "seconds")
````

## 8. Precedence and parentheses

**Precedence** determines which operation Python evaluates first. **Parentheses** make the intended order explicit.

### Practice: Predict both results.

````python
unit_price = 12.50
quantity = 4
shipping = 5

print(unit_price * quantity + shipping)
print(unit_price * (quantity + shipping))
````

### Precedence levels
Precedence levels determine the order of evaluation. Operations at the same level are evaluated from left to right:
  - Parentheses () Expressions within parentheses are evaluated first.
  - Exponentiation ```**```
  - Multiplication ```*```, Division ```/```, Floor Division ```//```, Modulus ```%```
  - Addition ```+```, Subtraction ```-```
  - Arithmetic built-ins: `abs`, `round`, `min`, `max`.



### Intermediate variables and code clarity
Intermediate variables can make the business meaning easier to review. Example: calculate the subtotal and total for an order.

````python
subtotal = unit_price * quantity
total = subtotal + shipping

print("Subtotal:", subtotal)
print("Total:", total)
````

## Scenario — A monthly savings plan

You want to study how a savings balance changes over several months. A single calculation gives one future balance, but a loop can repeat the calculation and a list can preserve the complete history.

We will build the program progressively and finish with a chart that makes the trend easier to interpret.

## 1. Lists preserve an ordered collection

A **list** stores several values in a defined order. A list is an **iterable** because Python can visit its elements one after another.

Positions are called **indices** and start at `0`. Therefore, a list with four elements has indices `0`, `1`, `2` and `3`.

````python
monthly_contributions = [100, 120, 80, 150]

print(monthly_contributions)
print("Number of contributions:", len(monthly_contributions))
print("First contribution:", monthly_contributions[0])
print("Fourth contribution:", monthly_contributions[3])
````

## 2. Visit every value with a `for` loop

A **loop** repeats a block of instructions. During each **iteration**, the loop variable receives the next value from the iterable. The indented block is the loop body.

Read the next loop as: *for each contribution in monthly contributions, display the contribution*.

````python
for contribution in monthly_contributions:
    print("Contribution:", contribution)

print("The loop has finished.")
````

## 3. Generate integer sequences with `range()`

`range()` produces an iterable sequence of integers. The stop value is **excluded**.

- `range(stop)` starts at `0`;
- `range(start, stop)` chooses the start;
- `range(start, stop, step)` also chooses the increment.

````python
print("range(4):")
for number in range(4):
    print(number)

print("range(1, 5):")
for number in range(1, 5):
    print(number)

print("range(2, 9, 2):")
for number in range(2, 9, 2):
    print(number)
````

## 4. Update program state in every iteration

A variable can preserve the current **state** of a calculation. Here, `balance` begins with an initial value and changes once per month.

Predict the final balance and trace the value after each iteration.

````python
initial_balance = 500
monthly_contribution = 100
number_of_months = 6
balance = initial_balance

for month in range(1, number_of_months + 1):
    balance = balance + monthly_contribution
    print("Month", month, "-> balance:", balance)

print("Final balance:", balance)
````

### Explain the state change

1. Why does `range(1, number_of_months + 1)` include month 6?
2. What would happen if `balance = initial_balance` were indented inside the loop?
3. At the end, does `balance` contain one value or the complete history?

## 5. Preserve the history with `append()`

The previous loop remembers only the latest balance. To analyse the evolution, we need a list.

The method `list.append(value)` adds one value to the end of an existing list. We will preserve the starting point as month `0`, then append every new month and balance.

````python
month_numbers = []
balances = []
balance = initial_balance

month_numbers.append(0)
balances.append(balance)

for month in range(1, number_of_months + 1):
    balance = balance + monthly_contribution
    month_numbers.append(month)
    balances.append(balance)

print("Months:", month_numbers)
print("Balances:", balances)
````

## 6. Use indices when the position matters

Direct iteration is ideal when we only need each value. An index-based loop is useful when we must connect corresponding positions in two lists.

`range(len(balances))` generates every valid non-negative index: from `0` to `len(balances) - 1`.

````python
for index in range(len(balances)):
    print("Position", index, "-> month", month_numbers[index], "-> balance", balances[index])
````

## 7. Preserve the indices that satisfy a condition

An index can tell us **where** an important event occurred, not only its value. The next loop stores all positions where the savings target has been reached.

````python
target_balance = 900
target_indices = []

for index in range(len(balances)):
    if balances[index] >= target_balance:
        target_indices.append(index)

print("Indices at or above the target:", target_indices)

if len(target_indices) > 0:
    first_target_index = target_indices[0]
    print("Target first reached in month:", month_numbers[first_target_index])
else:
    print("The target is not reached in this period.")
````

## 8. Represent the ordered series with Matplotlib

A chart helps us see a pattern across an ordered sequence. Matplotlib is a library, so we import the `pyplot` module before calling its functions.

The two plotted lists must have the same number of elements: each x-value needs one corresponding y-value.

````python
import matplotlib.pyplot as plt

plt.plot(month_numbers, balances, marker="o")
plt.xlabel("Month")
plt.ylabel("Balance (€)")
plt.title("Monthly savings plan")
plt.grid(True)
plt.show()
````

### Read the chart

## 9. Extension — Contributions can vary

A fixed contribution can be represented by one value. When each month has a different plan, a list is more appropriate.

Here, direct iteration gives us each contribution. We still preserve the resulting balance after every update.

````python
planned_contributions = [100, 120, 80, 150, 100, 130]
variable_balances = [initial_balance]
balance = initial_balance

for contribution in planned_contributions:
    balance = balance + contribution
    variable_balances.append(balance)

variable_months = []
for month in range(len(variable_balances)):
    variable_months.append(month)

print("Balances with a variable plan:", variable_balances)

plt.plot(variable_months, variable_balances, marker="o")
plt.xlabel("Month")
plt.ylabel("Balance (€)")
plt.title("Savings with variable contributions")
plt.grid(True)
plt.show()
````

## 10. Choose the loop pattern deliberately

| Need | Useful pattern |
|---|---|
| Use every value | `for value in values:` |
| Repeat a known number of times | `for step in range(number_of_steps):` |
| Connect values by position | `for index in range(len(values)):` |
| Preserve each result | create a list, then call `.append(...)` inside the loop |
| Show a trend | store the ordered values, then plot them |

Prefer direct iteration when the index adds no useful information.

## 11. Read and reason about common mistakes

### A. Off-by-one index

```python
for index in range(len(balances) + 1):
    print(balances[index])
```

The extra `+ 1` generates one invalid index.

### B. State reset inside the loop

```python
for month in range(1, 7):
    balance = initial_balance
    balance = balance + monthly_contribution
```

The previous month's state is lost in every iteration.

### C. Result not preserved

```python
for contribution in planned_contributions:
    balance = balance + contribution
variable_balances.append(balance)
```

Because `append()` is outside the loop, only the final balance is added.

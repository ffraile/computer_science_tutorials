## 1) Core Concepts: Numpy arrays
- NumPy arrays are **n-dimensional**, **homogeneous** data structures for numerical data.
- Arrays support **vectorized operations**: operations applied to entire arrays without explicit loops.
- **Array programming**: think in terms of whole arrays, not elementwise loops.
- Lots of built-in functions for math, stats, linear algebra, etc.
- Highly optimized for performance (C backend).

### 1.1 Creating arrays
- From Python objects: `np.array([1,2,3])`
- Ranges / grids: `np.arange`, `np.linspace`
- Filled arrays: `np.zeros`, `np.ones`, `np.full`, `np.eye`
- Random: Normal distribution `np.random.normal` (explore others)

````python

a = np.array([1, 2, 3], dtype=np.int32)
b = np.arange(0, 10, 2)           # step range
c = np.linspace(0, 1, 5)          # 5 points in [0,1]
Z = np.zeros((2,3))
O = np.ones((3,2))
F = np.full((2,2), 7.5)
I = np.eye(3)

R = np.random.normal(low=0, high=10, size=(2,3))
print("a:", a, a.dtype)
print("b:", b)
print("c:", c)
print("Z:", Z)
print("I:", I)
print("R:", R)
````

### 1.2 Basic Operations
- Elementwise operations: `+ - * / **`


````python
A = np.array([[1, 2, 3],
     [4, 5, 6]])

print("A:", A, sep="\n")
print("A + 10:", A + 10, sep="\n")
print("A * 2:", A * 2, sep="\n")
print("A ** 2:", A ** 2, sep="\n")  # make a copy to avoid modifying A

B = np.array([[10, 20, 30],
              [40, 50, 60]])

print("A + B:", A + B, sep="\n")
print("A * B:", A * B, sep="\n")
````

**!Important! tips**:
- 1) assigning parts of an array to another variable does not create a copy, it creates a **view**. Changes to the view will affect the original array. Use `.copy()` or `np.array(A)` to create a true copy.
- 2) Arrays are homogeneous: all elements have the same dtype.
- 3) Using `dtype` parameter when creating arrays ensures the desired data type (optimizes memory and performance).

## 2) Shape, dtype, size

Useful attributes:
- `.shape`: Shape of the array (eg 2x3)
- `.ndim`: Number of dimensions
- `.size`: Number of elements,
- `.dtype`: Type of elements,
- `.itemsize`: Size of elements in Memory
- `astype` to convert dtype

````python

M = np.random.normal(0, 100, size=(3,4))
print("shape:", M.shape, "| ndim:", M.ndim, "| size:", M.size)
print("dtype:", M.dtype, "| itemsize:", M.itemsize, "bytes")
print("astype float32:", M.astype(np.float32))
print("New itemsize: ", M.astype(np.float32).itemsize, "bytes")
````

## 3) Indexing & slicing

- 1D arrays indexing and slicing just like lists: `x[i]`, `x[i:j:k]`
- 2D arrays: row/column index/slices: `A[i, j]`, `A[i, j1:j2]`,

````python

x = np.arange(10)
A = np.random.normal(0, 20, size=(4,5))
print("x:", x)
print("x[2:8:2]:", x[2:8:2])
print("A:", A)
print("A[1, :]:", A[1, :]) # Second row
print("A[:, 2]:", A[:, 2]) # Third column
````

### 3.1 Boolean masks
Create a boolean mask and use it to filter an array.
- *Boolean mask* is an array of `True`/`False` values.


````python
# Filter data with boolean mask
print("A greater than 2:", A[A >= 2.0])

# We can store the boolean mask in a variable
mask = (A >= 2.0)
print("mask:", mask.astype(int))
print("A[mask]:", A[mask])
````

## 4) Broadcasting

Broadcasting allows operations on arrays of different shapes.
**Rules:** align shapes **from the right**; dims are compatible if equal or either is `1`.

Examples: add a vector to each row/column.

````python

B = np.random.normal(0, 10, size=(3,4))
row = np.array([1, 0, -1, 2])    # row vector shape (4,)
col = np.array([[10],[20],[30]]) # column vector shape (3,1)

print("B:", B)
print("B + row:", B + row)   # broadcast across rows
print("B + col:", B + col)   # broadcast down columns
````

## 5) Vectorized ops & ufuncs

- Elementwise: `+ - * / **`
- Ufuncs: `np.sqrt`, `np.exp`, `np.clip`, `np.where`
- Aggregations: `np.sum`, `np.mean`, `np.std`,
- `axis=` parameter to specify dimension

````python

X = np.linspace(-2, 2, 5)
print("X:", X)
print("X**2:", X**2)
print("sqrt(abs(X)):", np.sqrt(np.abs(X)))
print("where(X>=0, 1, -1):", np.where(X>=0, 1, -1))

G = np.random.normal(loc=0, scale=1, size=(3,4))
print("G:", G)
print("sum all:", np.sum(G))
print("mean per column:", np.mean(G, axis=0))
print("std per row:", np.std(G, axis=1))
````
